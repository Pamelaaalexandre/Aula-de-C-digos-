```python
# como funciona o while?
# while condição:

# Exemplo1 - contar até onde o usuario pedir
limite = int(input("Informe um numero limite: "))
inicio = 0

# O "=" significa "recebe" no TI.

while inicio <= limite:
  print(inicio)
  inicio = inicio + 1
```

```python
# controle com while
senha_correta= "abc123"
tentativas = 0

while tentativas < 3:
  senha = input("Digite sua senha: ")
  if senha_correta == senha:
    print("Bem vindo ao sistema")
    break #quebrar
  else:
    print("Senha incorreta")
    tentativas = tentativas + 1 #Em Python o correto é colocar tentativas +=1
    print(f"numero de tentativas = {tentativas}")
    if tentativas == 3:
      print("Senha bloqueada")
```

```python
# while true

while True:
  comando = input("Para sair, digite SAIR: ")
  if comando.lower() == "sair":
    print("Até mais")
    break
  else:
    print("continuando o sistema")
```

```python
# Atividade 1 - Facil
# Quero um sistema que pergunte ao usuario se ele quer salvar uma informação em uma lista de carros
# a lista-> carros = ["uno","corola", "fusca"]
# se o usuario quer colocare um novo carro, vai adicionar ele a lista
# digitar sair, finaliza o sistema

carros = ["Uno", "Corola", "Fusca"]
print("Bem Vindo ao sistema de cadastro de carros!")

while True:
  novo_carro = input("Deseja adicionar um novo carro ou digite sair para finalizar: ")
  if novo_carro.lower() == "sair":
    print("Até Mais")
    break
  else:
    carros.append(novo_carro)
    for i in carros:
      print(i)


```
